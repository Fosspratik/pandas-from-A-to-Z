```python
%pip install matplotlib
import pandas as pd 
import statsmodels.api as sm 
import matplotlib.pyplot as plt
dataframe= pd.DataFrame({"marks":[10,20,30,30,40],"hours":[2,4,1,2,3]})
x=dataframe["marks"]
y=dataframe["hours"]
z=sm.add_constant(y)
print(z)
dataframe.plot(kind="scatter",x="marks",y="hours")
lm=sm.OLS(y,x).fit()
print(lm)
```
